```python
env = simpy.Environment()
money = 1_000_000_00
num_fishes = 20
fish_size: Literal["Small_Fingerlin"] = "Small_Fingerlin"
duration = 30
seed = 30
timestamp = datetime.now()
```

```python
phases = [
    Phase(duration, [1], 0, "Large_Fingerlin"), 
    Phase(duration, [1], 0, "Juveniles"), 
    Phase(duration, [1], 0, "Medium_Table"), 
]

farm_model = FarmingModel(
    stocks=[
        Stock(id=1, type="Catfish", phases=phases, size=fish_size, count=num_fishes)
    ],
    feed_conversion_rate={0: 1.1, 200: 1.3},
    feed_floating={0: True, 200: False},
    capital=money
)

sim = Simulation(env, farms={"Random": farm_model})
```
