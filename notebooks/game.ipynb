```python
from kaggle_environments import make
from src.domains.player.player import Player, PlayerConfig, PrivateState
from src.models.game import RealityState
from src.models.player import ScoreWeights, ResourceWeights
```

```python
config = PlayerConfig(
    method="BASIC", 
    weight_learning_rate=.5,
    # score_weights=ScoreWeights(FUTURE_DISCOUNT_RATE=.5),
    # resource_weights=ResourceWeights(PRODUCE=2000)
)
```

```python
def agent(obs):
    global config
    private = PrivateState(
        config=config
    )

    p = Player(**(obs | {"private": private}))
    print(p.private.config.resource_weights)

    state = p.play().to_kaggle()
    config = p.private.config
    return state
```

```python
env = make("kaggriculture", configuration={"episodeSteps": 5}, debug=True)
_ = env.run([agent, "starter"]) 

# View result
final = env.steps[-1]
for i, s in enumerate(final):
    print(f"Player {i}: reward={s.reward}, status={s.status}")
```
