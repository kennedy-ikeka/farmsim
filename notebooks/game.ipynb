```python
from kaggle_environments import make
from src.domains.player.player import Player, PlayerConfig, PrivateState
from src.models.game import RealityState
from src.models.player import ScoreWeights, ResourceWeights
```

```python
def agent(obs):
    private = PrivateState(
        config=PlayerConfig(
            method="BASIC", 
            resource_weights=ResourceWeights(
                MONEY=2
            )
        )
    )

    p = Player(**(obs | {"private": private}))
    state = p.play().to_kaggle()
    return state
```

```python
env = make("kaggriculture", configuration={"episodeSteps": 720}, debug=True)
_ = env.run([agent, "starter"]) 

# View result
final = env.steps[-1]
for i, s in enumerate(final):
    print(f"Player {i}: reward={s.reward}, status={s.status}")
```
