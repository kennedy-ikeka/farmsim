```python
sim = Simulation()
pond = Pond(type="Earthen_Pond", sim=sim)
fish_ids = [Fish(type="Catfish", pond_id=pond.id, sim=sim).id for _ in range(100)]
farm = Farm(sim=sim, name="Random", pond_ids=[pond.id], active=False)
batch = Batch(sim=sim, farm_id=farm.id, pond_id=pond.id, fish_ids=fish_ids, cycle_id='1', duration=60)

sim.run(10)
```
